```python
import pandas as pd
from functions.eval import *
```

```python
model_pred_col = "Camelbert-MSA"
model_name = "CAMeL-Lab/bert-base-arabic-camelbert-msa-sentiment"
# model_name = "PRAli22/AraBert-Arabic-Sentiment-Analysis"
# model_pred_col = "AraBert"
```

```python
lime_eval = pd.read_csv("data/xai_eval/xai_eval_lime_" + model_pred_col + ".csv")
shap_eval = pd.read_csv("data/xai_eval/xai_eval_shap_" + model_pred_col + ".csv")
ig_eval = pd.read_csv("data/xai_eval/xai_eval_ig_" + model_pred_col + ".csv")
dl_eval = pd.read_csv("data/xai_eval/xai_eval_dl_" + model_pred_col + ".csv")
ensemble_eval = pd.read_csv("data/xai_eval/xai_eval_ensemble_" + model_pred_col + ".csv")
```

```python
eval_df = pd.concat([lime_eval, shap_eval, ig_eval, dl_eval, ensemble_eval], axis=1)
```

```python
XAI_methods = ["LIME", "SHAP", "IG", "DeepLIFT", "EXAI_LIME_SHAP_IG_DL_mean", 
               "EXAI_LIME_SHAP_IG_DL_median", "EXAI_LIME_SHAP_IG_mean", 
               "EXAI_LIME_SHAP_IG_median", "EXAI_LIME_SHAP_mean"]
metrics = ["comprehensiveness", "sufficiency", "corr_loo", "ins_AUC", "del_AUC", "combined"]
```

```python
df = pd.DataFrame(
    {metric: [] for metric in metrics},
)
```

```python
for xai_method in XAI_methods:
    for metric in metrics:
        col_name = f"{xai_method}_{metric}"
        if col_name in eval_df.columns:
            df.loc[xai_method, metric] = eval_df[col_name].mean()
        else:
            df.loc[xai_method, metric] = None
```

```python
borda_count(df)
```

```python
df_std = df.copy()
for xai_method in XAI_methods:
    for metric in metrics:
        col_name = f"{xai_method}_{metric}"
        if col_name in eval_df.columns:
            df_std.loc[xai_method, metric] = eval_df[col_name].std()
        else:
            df_std.loc[xai_method, metric] = None
```

```python
df_std.drop(columns=["borda"], inplace=True)
df_std
```

```python
df_std2 = df2.copy()
for xai_method in XAI_methods:
    for metric in metrics:
        col_name = f"{xai_method}_{metric}"
        if col_name in eval_df.columns:
            df_std2.loc[xai_method, metric] = eval_df[col_name].std()
        else:
            df_std2.loc[xai_method, metric] = None
```

```python
df_avg = (df + df2) / 2
borda_count(df_avg)
```
