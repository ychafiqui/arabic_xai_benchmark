```python
import pandas as pd
from functions.eval import *
```

```python
model_pred_col = "Camelbert-MSA"
model_name = "CAMeL-Lab/bert-base-arabic-camelbert-msa-sentiment"
# model_name = "PRAli22/AraBert-Arabic-Sentiment-Analysis"
# model_pred_col = "AraBert"
```

```python
eval_df = pd.read_csv("data/hard_rationale/hard_rationale_ensemble_" + model_pred_col + ".csv")
```

```python
hard_rationale_choices = ["elbow_simple-lmethod", "elbow_kneedle", "elbow_dfdt", "elbow_lmethod",
                          "top_n_3", "top_n_5", "top_n_10", "top_n_20", 
                          "threshold_0.1", "threshold_0.3", "threshold_0.5", "threshold_0.7"]
metrics = ["comprehensiveness", "sufficiency", "combined"]
```

```python
df = pd.DataFrame(
    {metric: [] for metric in metrics},
)
```

```python
for hr in hard_rationale_choices:
    for metric in metrics:
        col_name = f"EXAI_LIME_SHAP_mean_{hr}_{metric}"
        if col_name in eval_df.columns:
            df.loc[hr, metric] = eval_df[col_name].mean()
        else:
            df.loc[hr, metric] = None
```

```python
df = borda_count_hard_rationale(df)
```

```python
for hr in hard_rationale_choices:
    col_name = f"nb_selected_tokens_{hr}"
    col2_name = f"%_tokens_input_{hr}"
    if col_name in eval_df.columns:
        df.loc[hr, "avg_nb_selected_tokens"] = eval_df[col_name].mean()
    else:
        df.loc[hr, "avg_nb_selected_tokens"] = None
    if col2_name in eval_df.columns:
        df.loc[hr, "avg_%_tokens_input"] = eval_df[col2_name].mean()
    else:
        df.loc[hr, "avg_%_tokens_input"] = None
df
```

```python
df_std = df.copy()
for hr in hard_rationale_choices:
    for metric in metrics:
        col_name = f"EXAI_LIME_SHAP_mean_{hr}_{metric}"
        if col_name in eval_df.columns:
            df_std.loc[hr, metric] = eval_df[col_name].std()
        else:
            df_std.loc[hr, metric] = None
df_std.drop(columns=["borda"], inplace=True)
for hr in hard_rationale_choices:
    col_name = f"nb_selected_tokens_{hr}"
    col2_name = f"%_tokens_input_{hr}"
    if col_name in eval_df.columns:
        df_std.loc[hr, "avg_nb_selected_tokens"] = eval_df[col_name].std()
    else:
        df_std.loc[hr, "avg_nb_selected_tokens"] = None
    if col2_name in eval_df.columns:
        df_std.loc[hr, "avg_%_tokens_input"] = eval_df[col2_name].std()
    else:
        df_std.loc[hr, "avg_%_tokens_input"] = None
df_std
```
